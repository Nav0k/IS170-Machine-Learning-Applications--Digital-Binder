```python
# call the required libraries
import pandas as pd
import numpy as np
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
#Importing dataset into colab
df=pd.read_csv('gamestore.csv')
df.head()
```

```python
#Adding prodcut as column header
df=pd.read_csv('gamestore.csv', names=['Product'])
df.head()
```

```python
#Code to show shape of the dataset showing 400 rows with one column
df.shape
```

```python
#converting pandas dataframe into a list of lists
data=list(df["Product"].apply(lambda x:x.split(",")))
data
```

```python
print(df.isnull().sum())  # Counts the number of missing (NaN) values in each column
```

```python
#transform the list with one-hot encoding
from mlxtend.preprocessing import TransactionEncoder
a=TransactionEncoder()
a_data=a.fit(data).transform(data)
df=pd.DataFrame(a_data, columns=a.columns_)

#Turn False & True to 0 and 1 for readability
df=df.replace(False,0)
df=df.replace(True,1)

df
```

```python
# SUPPORT threshold value
df=apriori(df,min_support=0.1,use_colnames=True)
df
```

```python
#Using association_rules for the confidence parameter
df_ar=association_rules(df,metric="confidence", min_threshold=0.6)
df_ar
```

```python
#Here is the LIFT parameter.
df_ar=association_rules(df,metric="lift", min_threshold=1)
df_ar
```

```python
#Here is the CONVICTION parameter.
df_ar=association_rules(df,metric="conviction", min_threshold=1.0)
df_ar
```

The association rule analysis identified several meaningful purchasing patternsw ithin the Gamestore dataset. The strongest rule was Batteries -> Gift Card, with a support of 0.16, confidence of 0.438, and lift of 1.103. This means that 16% of all transactions contained both items, and when Batteries were purchaced, there was a 43.8% chance that a Gift Card was also included. The lieft value greater than 1 indicated a positive relationship, meaning customers who buy Batteries are about 10% more likely to also purchase a Gift Card compared to random chance. Other notable patterns include Console -> Controller (confidence around 39.7%) and Console -> Headset (confidence around 36.2%), suggesting that accessories are commonly purchased alongside consoles.

Overall, the lift values across the rules are slightly above 1, which confirms positive but moderate associations between products. While the relationships are not extremely strong, they stulle provide useful business insight. These patterns suggest opportunities for bundling consoles with controllers or headset, recommending batteries with related accessories, and strategically placing complementary products near each other in-store or online. Leveraging these associations could help increase average transaction value and improve cross-selling effectiveness.
